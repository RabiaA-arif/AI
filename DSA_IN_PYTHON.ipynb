```python
# password match
import string
from itertools import product

def brute_force_password_match(password,max_length):
  char = string.ascii_lowercase
  attempts = 0
  for lenght in range(1,max_length + 1):
    for attempt in product(char,repeat=lenght):
      attempts +=1
      candidate = "".join(attempt)
      if candidate == password:
        return candidate , attempts

      return None,attempts

password:str = "yes65432!@"
# print(len(password))
max_len=len(password)
match ,total_attempts = brute_force_password_match(password,max_len)

if match:
  print(f"password is {match} after {total_attempts} : attemps")

else:
  print(f"password is not found within {total_attempts} attempts")

```

```python
import statistics

def array_median(num1: list[int], num2: list[int]) -> float:
  merged = num1 + num2
  merged.sort()
  mediaan = statistics.median(merged)
  print(f"Merged Of Two Array:{merged}")
  print(f"Median of:{mediaan}")

obj = array_median([2,3],[1,5])
obj1 = array_median([3,4,5],[1])
```

```python
class Solution:
  def find_median_sorted_array(self, num1: list[int], num2: list[int]):
    merged = []
    i = 0
    j = 0

    while i < len(num1) and j < len(num2):
      if num1[i] < num2[j]:
        merged.append(num1[i])
        i+=1
      else:
        merged.append(num2[j])
        j+=1
    print(f"Merged Array:{merged}")

    while i < len(num1):
      merged.append(num1[i])
      i+=1

    while j < len(num2):
      merged.append(num2[j])
      j+=1
    print(f"Merged Array:{merged}")

    mid = len(merged) // 2
    if len(merged) % 2 ==0:
      return (merged[mid -1] + merged[mid] /2)

    else:
      return merged[mid]


obj = Solution()
obj.find_median_sorted_array([2,3,1],[7,3])
```

```python
class Solution:
    def findMedianSortedArrays(self, nums1: list[int], nums2: list[int])-> float:
        merged = []
        i, j = 0, 0

        while i < len(nums1) and j < len(nums2):
            if nums1[i] < nums2[j]:
                merged.append(nums1[i])
                i += 1
            else:
                merged.append(nums2[j])
                j += 1

        while i < len(nums1):
            merged.append(nums1[i])
            i += 1

        while j < len(nums2):
            merged.append(nums2[j])
            j += 1

        mid = len(merged) // 2
        if len(merged) % 2 == 0:
            return (merged[mid-1] + merged[mid]) / 2
        else:
            return merged[mid]

```

## **Recursion makes the stack visible**


The printed stack will list, top to bottom: print_stack is called from countdown(0), which was called from countdown(1), from countdown(2), from countdown(3), from main, from module top-level. Each recursive call is one frame. This is exactly why deep recursion eventually overflows the stack — every call adds a frame and nothing is popped until the base case is hit.

```python
import traceback
def countdown(n):
  if n == 0:
    traceback.print_stack()
    return
  countdown(n-1)

def main():
  countdown(3)

main()
```

## **A frame holds locals — and they vanish on**

```python
#include <stdio.h>

int* dangerous(void) {
    int local = 42;
    return &local;   // BUG: returning a pointer into our own frame
}                    // <- frame is abandoned here; `local` is gone

int main(void) {
    int *p = dangerous();
    // *p now points into a frame that no longer exists.
    // The memory may still hold 42... or garbage... or the next call's data.
    printf("%d\n", *p);   // undefined behavior
    return 0;
}
```

## **Palindrome String**

a palindrome string is a sequence of characters that reads the exact same forward and backward

```python
# two pointer method
def ispalindrome(s):
  first = 0
  last = len(s) -1
  print(last)

  while first < last:
    if s[first] != s[last]:
      return False
    first+=1
    last-=1

  return True

obj = ispalindrome("mamm")
print(obj)
```

```python
# using single variable
def ispalindrome(s):
  length = len(s)
  for i in range(length //2):
    if s[i] != s[length - i-1]:
      return False
  return True

obj1 = ispalindrome("mam")
print(obj1)
```

```python
# Longest palindrome substrings
def get_longest_palindrome(s):
  n = len(s)
  print(f"Length of string:{n}")
  dp = [[False] *n for _ in range(n)]
  # print(f"Value of dp:{dp}")
  start = 0
  maxLen = 1
  for i in range(n):
    dp[i][i] = True
    # print(dp[i][i])
  for i in range(n-1):
    if s[i] == s[i+1]:
      dp[i][i+1] = True
      if maxLen == 1:
        start = i
        maxLen = 2
  for length in range(3,n+1):
    for i in range(n- length+1):
      j = i+length -1
      if s[i] == s[j] and dp[i+1][j-1]:
        dp[i][j] = True
        if length > maxLen:
          start = i
          maxLen = length
  return s[start:start + maxLen]
if __name__ == "__main__":
  s = "hellohhhbballeo"
  print(get_longest_palindrome(s))
```

```python
# Mean of Array
# Given an unsorted array arr[], the task is to find the mean of the array.
# Note: Return the floor value of the mean.
# my approach
def array_mean(arr):
  mean_arr = sum(arr)//len(arr)
  return mean_arr

array_mean([1, 3, 4, 2, 6, 5, 8, 7])
array_mean([4, 4, 4, 4, 4])
```

```python
# Naive approach  O(n) Time O(1) Space
# The idea is to traverse the array to compute the sum,
# then divide it by n to get the average.
# Since the return type is int, the result is the floor of the mean.

def calcu_mean(arr):
  sum = 0
  arr_len = len(arr)

  for i in arr:
    sum+=i
  result = int(sum/arr_len)
  # print(sum)
  return result


calcu_mean([1, 3, 4, 2, 6, 5, 8, 7])
calcu_mean([4, 4, 4, 4, 4])



```
