```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
# define text and from that text create set of characters
text = "I am a perfect developer mindset and love to teach machines"
chars = sorted(list(set(text)))  # IN RNN text oder and context matter both ,# define text and using set we identify unique charater text
# chars
chars_to_index = {char:i for i ,char in enumerate(chars)}
chars_to_index
index_to_character = {i: char for i,char in enumerate(chars)}
# index_to_character
```

```python
# create sequence and labels of text
seqence_length = 3
sequence = []
labels = []
# print(len(text))

for i in range(len(text) - seqence_length):
  # print(i)
  seq = text[i:i + seqence_length]
  label = text[i + seqence_length]
  sequence.append([chars_to_index[char] for char in seq])
  labels.append(chars_to_index[label])

  # print(labels)
  # print(seq)
# print(seqence_length)
# print(sequence)
X = np.array(sequence)
y = np.array(labels)
```
