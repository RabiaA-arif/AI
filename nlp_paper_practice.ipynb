## RNN

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
```

```python
# define text and from that text create set of characters
text = "I am a perfect developer mindset and love to teach machines,I love programming from deep heartedly"
chars = sorted(list(set(text)))  # IN RNN text oder and context matter both ,# define text and using set we identify unique charater text
# chars
chars_to_index = {char:i for i ,char in enumerate(chars)}
chars_to_index
index_to_character = {i: char for i,char in enumerate(chars)}
# index_to_character
```

```python
# create sequence and labels of text
seqence_length = 3
sequence = []
labels = []
# print(len(text))

for i in range(len(text) - seqence_length):
  # print(i)
  seq = text[i:i + seqence_length]
  label = text[i + seqence_length]
  sequence.append([chars_to_index[char] for char in seq])
  labels.append(chars_to_index[label])

  # print(labels)
  # print(seq)
# print(seqence_length)
# print(sequence)
X = np.array(sequence)
y = np.array(labels)
```

```python
# convert seuence and labels to one hot coding
x_onehot = tf.one_hot(X, len(chars))
y_onehot = tf.one_hot(y, len(chars))
```

```python
# creating RNN model
model = Sequential()
model.add(SimpleRNN(50, input_shape=(seqence_length , len(chars)),activation='relu'))
model.add(Dense(len(chars), activation='softmax'))
```

```python
# compiling and training the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(x_onehot, y_onehot, epochs=250)
```

```python
# generating new text using our trained model
text_strat = 'i am a perfect'
genrated_txt = text_strat
for i in range(70):
  x = np.array([[chars_to_index[char] for char in genrated_txt[-seqence_length:]]])
  x_onehot = tf.one_hot(x, len(chars))
  prediction = model.predict(x_onehot)
  next_index = np.argmax(prediction)
  next_char = index_to_character[next_index]
  genrated_txt  += next_char

print("RNN Generated Text")
print(genrated_txt)

```
